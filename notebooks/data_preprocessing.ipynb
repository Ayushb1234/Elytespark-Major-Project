```python
import pandas as pd
import numpy as np


class DataCleaner:

    def __init__(self, df):
        self.df = df.copy()

    # -------------------------------------------------
    # 1. Standardize column names
    # -------------------------------------------------
    def standardize_columns(self):

        self.df.columns = (
            self.df.columns
            .str.strip()
            .str.lower()
            .str.replace(" ", "_")
        )

        return self


    # -------------------------------------------------
    # 2. Remove duplicate records
    # -------------------------------------------------
    def remove_duplicates(self):

        before = len(self.df)

        self.df.drop_duplicates(inplace=True)

        after = len(self.df)

        print(f"Duplicate rows removed: {before - after}")

        return self


    # -------------------------------------------------
    # 3. Remove identifier
    # -------------------------------------------------
    def drop_identifier(self):

        self.df.drop(
            columns=["composite_key"],
            errors="ignore",
            inplace=True
        )

        return self


    # -------------------------------------------------
    # 4. Standardize categorical text
    # -------------------------------------------------
    def clean_text_columns(self):

        cols = self.df.select_dtypes(include=["object", "string"]).columns

        for col in cols:

            self.df[col] = (
                self.df[col]
                .astype("string")
                .str.strip()
                .str.lower()
            )

        return self


    # -------------------------------------------------
    # 5. Missing values
    # -------------------------------------------------
    def handle_missing_values(self):

        numeric_cols = self.df.select_dtypes(
            include=np.number
        ).columns

        categorical_cols = self.df.select_dtypes(
            include=["object", "string", "category"]
        ).columns

        # Numerical -> Median
        for col in numeric_cols:

            self.df[col] = self.df[col].fillna(
                self.df[col].median()
            )

        # Categorical -> Mode
        for col in categorical_cols:

            mode = self.df[col].mode(dropna=True)

            if not mode.empty:
                self.df[col] = self.df[col].fillna(mode.iloc[0])

        return self


    # -------------------------------------------------
    # 6. Convert numeric-looking columns
    # -------------------------------------------------
    def convert_numeric_columns(self):

        numeric_cols = [
            "age",
            "age_normalized",
            "bmi",
            "hba1c_level",
            "glucose",
            "cholesterol",
            "sleep_hours",
            "triglycerides",
            "crp_level",
            "homocysteine_level",
            "systolic_bp",
            "diastolic_bp",
            "alcohol_intake",
            "salt_intake",
            "heart_rate",
            "hdl",
            "ldl"
        ]

        for col in numeric_cols:

            if col in self.df.columns:

                self.df[col] = pd.to_numeric(
                    self.df[col],
                    errors="coerce"
                )

        return self


    # -------------------------------------------------
    # 7. Binary Encoding
    # -------------------------------------------------
    def encode_binary_columns(self):

        binary_mapping = {
            "yes": 1,
            "no": 0,

            "true": 1,
            "false": 0,

            "positive": 1,
            "negative": 0,

            "male": 1,
            "female": 0
        }

        binary_cols = [
            "gender",
            "smoking",
            "diabetes",
            "hypertension",
            "heart_disease",
            "family_history",
            "low_hdl_cholesterol",
            "high_ldl_cholesterol",
            "high_blood_pressure"
        ]

        for col in binary_cols:

            if col in self.df.columns:

                # Only mapping strings.
                # Existing 0/1 values remain unchanged.
                if not pd.api.types.is_numeric_dtype(self.df[col]):

                    self.df[col] = (
                        self.df[col]
                        .astype("string")
                        .str.strip()
                        .str.lower()
                        .replace(binary_mapping)
                    )

                    self.df[col] = pd.to_numeric(
                        self.df[col],
                        errors="coerce"
                    )

        return self


    # -------------------------------------------------
    # 8. Ordinal Encoding
    # -------------------------------------------------
    def encode_ordinal_columns(self):

        # Age category
        age_mapping = {
            "young": 0,
            "adult": 1,
            "middle_age": 2,
            "middle aged": 2,
            "middle-aged": 2,
            "senior": 3,
            "elderly": 3
        }

        # BMI category
        bmi_mapping = {
            "underweight": 0,
            "normal": 1,
            "normal weight": 1,
            "overweight": 2,
            "obese": 3,
            "obesity": 3
        }

        # Stress
        stress_mapping = {
            "low": 0,
            "medium": 1,
            "moderate": 1,
            "high": 2
        }

        # Physical activity
        activity_mapping = {
            "low": 0,
            "medium": 1,
            "moderate": 1,
            "high": 2
        }

        # Blood pressure category
        bp_mapping = {
            "normal": 0,
            "elevated": 1,
            "high": 2,
            "hypertension": 2
        }

        mappings = {
            "age_level": age_mapping,
            "bmi_level": bmi_mapping,
            "stress_level": stress_mapping,
            "physical_activity": activity_mapping,
            "blood_pressure": bp_mapping
        }

        for col, mapping in mappings.items():

            if col in self.df.columns:

                # Don't touch already numerical encoding
                if not pd.api.types.is_numeric_dtype(self.df[col]):

                    self.df[col] = (
                        self.df[col]
                        .astype("string")
                        .str.strip()
                        .str.lower()
                        .map(mapping)
                    )

        return self


    # -------------------------------------------------
    # 9. Validate Encoding
    # -------------------------------------------------
    def check_remaining_categories(self):

        categorical_cols = self.df.select_dtypes(
            include=["object", "string", "category"]
        ).columns.tolist()

        print("\nRemaining categorical columns:")

        if categorical_cols:

            for col in categorical_cols:

                print(f"\n{col}")

                print(
                    self.df[col]
                    .value_counts(dropna=False)
                    .head(20)
                )

        else:

            print("No categorical columns remaining.")

        return self


    # -------------------------------------------------
    # Complete cleaning pipeline
    # -------------------------------------------------
    def clean(self):

        self.standardize_columns()

        self.remove_duplicates()

        self.drop_identifier()

        self.clean_text_columns()

        self.convert_numeric_columns()

        self.encode_binary_columns()

        self.encode_ordinal_columns()

        # Run after conversion/encoding because those operations
        # may create NaN for invalid values.
        self.handle_missing_values()

        self.check_remaining_categories()

        return self.df
```
